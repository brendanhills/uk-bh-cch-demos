```python
import re

def find_int_words(line):
  word_index = {}
  word_list = [["one", "1"], ["two", "2"], ["three", "3"], ["four", "4"], ["five", "5"], ["six", "6"], ["seven", "7"], ["eight", "8"], ["nine", "9"]]

  for num_word in enumerate(word_list):
    for num in num_word:
      #word_location = line.find(num)
      word_matches = re.finditer(num, line)
      for word_match in word_matches:
        num = num_word[1]
        word_location = word_match.start()
        word_index[word_location] =int(num)
        print(f"{num=}, at {word_location=}, {num=}, ")

  return word_index

def get_first_and_last(words_dict):
  return [words_dict[min(words_dict)], words_dict[max(words_dict)]]




def get_calibration2(text):
  calibration_value = 0
  lines=text.splitlines()
  print(f"{lines=}")
  for line in lines:
    print(f"{line=}")
    numbers = find_int_words(line)
    first_and_last = get_first_and_last(numbers)
    line_val = 10*first_and_last[0] + first_and_last[1]
    print(f"{numbers=} {first_and_last=} {line_val=}")
    calibration_value += line_val
    print(f"Calibration: {calibration_value=}")
  return calibration_value

get_calibration2(sample_text)
```
